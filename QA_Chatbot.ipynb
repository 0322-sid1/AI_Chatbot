```python
# CELL 3: Text Preprocessing Pipeline
# Steps: lowercase → remove special chars → remove stopwords → lemmatize
import re
import string
import pandas as pd
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# ── Initialize NLP tools
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words("english"))

# Keep question words — removing them breaks Q&A matching
KEEP_WORDS = {"what","who","where","when","why","how","which","is","are","does"}
stop_words = stop_words - KEEP_WORDS


def preprocess(text: str) -> str:
    """
    Full text preprocessing pipeline:
      1. Lowercase
      2. Remove punctuation and special characters
      3. Remove extra whitespace
      4. Remove stopwords (keeping question words)
      5. Lemmatize tokens

    Example:
      Input : "What is Machine Learning?"
      Output: "what is machine learning"
    """
    if not text or not isinstance(text, str):
        return ""

    # 1. Lowercase
    text = text.lower()

    # 2. Remove special characters (keep letters and spaces)
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # 3. Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # 4. Tokenize, remove stopwords, lemmatize
    tokens = [
        lemmatizer.lemmatize(word)
        for word in text.split()
        if word not in stop_words and len(word) > 1
    ]

    return " ".join(tokens)


# ── Apply preprocessing to dataset
df = pd.read_csv("qa_dataset.csv")
df["clean_question"] = df["question"].apply(preprocess)

# Show comparison
print("✅ Preprocessing applied!\n")
print("Before → After examples:")
print("─" * 65)
for i in [0, 25, 50, 75, 100]:
    orig  = df.iloc[i]["question"]
    clean = df.iloc[i]["clean_question"]
    print(f"  Original : {orig}")
    print(f"  Cleaned  : {clean}")
    print()
```

```python
# CELL 4: Build TF-IDF Matrix and Retrieval Engine
# Uses n-grams (1,2) for better matching of multi-word phrases
import numpy as np
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Build TF-IDF Matrix
vectorizer = TfidfVectorizer(
    ngram_range = (1, 2),    # unigrams + bigrams ("machine learning" as one feature)
    max_features= 5000,       # top 5000 most important terms
    analyzer    = "word",
    min_df      = 1,          # include terms appearing even once
    sublinear_tf= True,       # apply log scaling to term frequency
)

# Fit on all cleaned questions
X = vectorizer.fit_transform(df["clean_question"])

print(f"✅ TF-IDF Matrix built!")
print(f"   Shape       : {X.shape}  ({X.shape[0]} questions × {X.shape[1]} features)")
print(f"   N-grams     : (1, 2) — unigrams + bigrams")
print(f"   Total features: {len(vectorizer.get_feature_names_out())}")

# ── Save model to disk
with open("tfidf_model.pkl", "wb") as f:
    pickle.dump({"vectorizer": vectorizer, "X": X, "df": df}, f)
print("   Model saved → tfidf_model.pkl")

# ── Show sample TF-IDF features
features = vectorizer.get_feature_names_out()
print(f"\n  Sample features: {list(features[:10])} ...")
print(f"  Sample bigrams : {[f for f in features if ' ' in f][:5]}")
```

```python
# CELL 5: Chatbot function with:
#   ✅ Similarity threshold (returns sorry if no good match)
#   ✅ Shows matched question (transparency)
#   ✅ Shows similarity score
#   ✅ Domain filter option
#   ✅ Top-N results option

# ── Similarity threshold
THRESHOLD = 0.15   # below this → "I don't understand"
                   # TF-IDF scores are often lower than you'd expect
                   # 0.15 works well for this dataset size


def get_recommendations(
    query         : str,
    domain_filter : str  = None,
    top_n         : int  = 3,
) -> list:
    """
    Find top-N most similar Q&A pairs for a user query.

    Args:
        query        : user's question
        domain_filter: optional — filter to one domain only
        top_n        : number of results to return

    Returns:
        List of dicts with keys:
          rank, question, answer, domain, similarity_score
    """
    # Step 1: Preprocess the query
    clean_query = preprocess(query)
    if not clean_query:
        return []

    # Step 2: Vectorize the query using fitted TF-IDF
    query_vec = vectorizer.transform([clean_query])

    # Step 3: Compute cosine similarity against all questions
    if domain_filter and domain_filter != "All":
        # Filter dataset to selected domain only
        mask        = df["domain"] == domain_filter
        filtered_df = df[mask].reset_index(drop=True)
        filtered_X  = X[mask.values]
        scores      = cosine_similarity(query_vec, filtered_X)[0]
        search_df   = filtered_df
    else:
        scores    = cosine_similarity(query_vec, X)[0]
        search_df = df

    # Step 4: Get top-N indices sorted by score
    top_indices = np.argsort(scores)[::-1][:top_n]

    # Step 5: Build results
    results = []
    for rank, idx in enumerate(top_indices, start=1):
        score = float(scores[idx])
        if score < THRESHOLD:
            break   # stop if below threshold
        row = search_df.iloc[idx]
        results.append({
            "rank"             : rank,
            "question"         : row["question"],
            "answer"           : row["answer"],
            "domain"           : row["domain"],
            "similarity_score" : round(score, 4),
        })

    return results


def chatbot(
    query         : str,
    domain_filter : str = None,
    verbose       : bool = True,
) -> str:
    """
    Main chatbot function.

    Args:
        query        : user's input question
        domain_filter: optional domain to search within
        verbose      : if True, prints matched question and score

    Returns:
        Best matching answer string, or "I don't understand" message.
    """
    if not query.strip():
        return "Please ask a question!"

    results = get_recommendations(query, domain_filter, top_n=1)

    if not results:
        return (
            "❓ Sorry, I don't have enough information to answer that.\n"
            "   Try rephrasing your question or ask about:\n"
            "   AI, Python, Health, Education, or Science."
        )

    best = results[0]

    if verbose:
        print(f"  📌 Matched : '{best['question']}'")
        print(f"  🏷️  Domain  : {best['domain']}")
        print(f"  📊 Score   : {best['similarity_score']:.4f}")
        print(f"  {'─'*50}")

    return best["answer"]


# ── Quick functionality test
print("✅ Chatbot function ready!\n")
print("Test queries:")
print("=" * 55)
tests = [
    ("what is machine learning",    None),
    ("explain python programming",  None),
    ("how much water to drink",     "Health & Medicine"),
    ("how to study effectively",    "Education"),
    ("what is a black hole",        "General Science"),
    ("random gibberish xyzabc",     None),   # should trigger threshold
]
for query, domain in tests:
    print(f"\n  You [{domain or 'All'}]: {query}")
    resp = chatbot(query, domain_filter=domain, verbose=True)
    print(f"  Bot: {resp[:80]}...")
```

```python
# CELL 6: Interactive chatbot in terminal mode
# Type your question, type 'exit' to quit
# Type 'domains' to see available domains
# Type 'domain:Health' to filter to a specific domain
DOMAINS = ["All"] + sorted(df["domain"].unique().tolist())

print("=" * 58)
print("  🤖 Domain-Specific Q&A Chatbot — Terminal Mode")
print("=" * 58)
print("  Commands:")
print("    'domains'       → show all available domains")
print("    'domain:Health' → filter answers to Health only")
print("    'top3'          → show top 3 matching answers")
print("    'exit'          → quit")
print("=" * 58)
print()

current_domain = None   # None = search all domains
show_top       = 1      # how many results to show

while True:
    try:
        user_input = input("You: ").strip()
    except (EOFError, KeyboardInterrupt):
        print("\nBot: Goodbye! 👋")
        break

    if not user_input:
        continue

    # ── Commands
    if user_input.lower() == "exit":
        print("Bot: Goodbye! Thanks for using Nova Q&A Bot. 👋")
        break

    elif user_input.lower() == "domains":
        print(f"Bot: Available domains: {', '.join(DOMAINS)}")
        continue

    elif user_input.lower().startswith("domain:"):
        requested = user_input.split(":", 1)[1].strip()
        if requested in DOMAINS:
            current_domain = None if requested == "All" else requested
            print(f"Bot: ✅ Domain filter set to: '{requested}'")
        else:
            print(f"Bot: ❌ Unknown domain. Choose from: {', '.join(DOMAINS)}")
        continue

    elif user_input.lower() == "top3":
        show_top = 3
        print("Bot: ✅ Will now show top 3 matches.")
        continue

    elif user_input.lower() == "top1":
        show_top = 1
        print("Bot: ✅ Will now show top 1 match.")
        continue

    # ── Normal Q&A
    if show_top == 1:
        response = chatbot(user_input, domain_filter=current_domain, verbose=True)
        print(f"Bot: {response}\n")
    else:
        results = get_recommendations(user_input, current_domain, top_n=3)
        if not results:
            print("Bot: ❓ Sorry, I don't understand. Try rephrasing.\n")
        else:
            print(f"Bot: Found {len(results)} relevant answers:\n")
            for r in results:
                print(f"  [{r['rank']}] Score: {r['similarity_score']:.4f} | {r['domain']}")
                print(f"       Q: {r['question']}")
                print(f"       A: {r['answer'][:100]}...")
                print()
```

```python
# CELL 7: Write Streamlit UI
%%writefile streamlit_app.py
"""
streamlit_app.py
================
Streamlit web UI for the Domain-Specific Q&A Chatbot.

Features:
  - Chat interface
  - Domain filter dropdown
  - Top-N results slider
  - Similarity score bar
  - Matched question highlighted
  - Conversation history
  - Dataset explorer tab
"""

import re
import pickle
import string
import numpy as np
import pandas as pd
import streamlit as st
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.metrics.pairwise import cosine_similarity
import nltk

nltk.download("stopwords", quiet=True)
nltk.download("wordnet",   quiet=True)

# ── Page Config
st.set_page_config(
    page_title = "Q&A Chatbot",
    page_icon  = "🤖",
    layout     = "wide",
)

# ── CSS
st.markdown("""
<style>
.user-bubble {
    background: linear-gradient(135deg, #667eea, #764ba2);
    color: white; padding: 12px 18px;
    border-radius: 18px 18px 4px 18px;
    margin: 8px 0 8px 80px; font-size: 15px;
}
.bot-bubble {
    background: #1e2130; color: #e8e8e8;
    padding: 14px 18px;
    border-radius: 18px 18px 18px 4px;
    margin: 8px 80px 8px 0; font-size: 15px;
    border: 1px solid #2d3561;
}
.match-box {
    background: #0d2137; border-left: 4px solid #667eea;
    padding: 10px 14px; border-radius: 0 8px 8px 0;
    margin: 6px 0; font-size: 13px; color: #aac4e8;
}
.domain-badge {
    display: inline-block;
    background: #2d3561; color: #aac4e8;
    padding: 2px 10px; border-radius: 12px;
    font-size: 12px; margin-left: 6px;
}
</style>
""", unsafe_allow_html=True)


# ── Load model and data
@st.cache_resource
def load_model():
    with open("tfidf_model.pkl", "rb") as f:
        payload = pickle.load(f)
    return payload["vectorizer"], payload["X"], payload["df"]

vectorizer, X, df = load_model()
DOMAINS = ["All"] + sorted(df["domain"].unique().tolist())
THRESHOLD = 0.15


# ── Preprocessing
@st.cache_resource
def get_nlp_tools():
    lemmatizer = WordNetLemmatizer()
    keep_words = {"what","who","where","when","why","how","which","is","are","does"}
    stops      = set(stopwords.words("english")) - keep_words
    return lemmatizer, stops

lemmatizer, stop_words = get_nlp_tools()

def preprocess(text):
    if not text: return ""
    text = text.lower()
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    tokens = [lemmatizer.lemmatize(w) for w in text.split()
              if w not in stop_words and len(w) > 1]
    return " ".join(tokens)


def get_recommendations(query, domain_filter=None, top_n=3):
    clean_q   = preprocess(query)
    if not clean_q:
        return []

    query_vec = vectorizer.transform([clean_q])

    if domain_filter and domain_filter != "All":
        mask       = df["domain"] == domain_filter
        search_df  = df[mask].reset_index(drop=True)
        search_X   = X[mask.values]
    else:
        search_df = df
        search_X  = X

    scores      = cosine_similarity(query_vec, search_X)[0]
    top_indices = np.argsort(scores)[::-1][:top_n]

    results = []
    for rank, idx in enumerate(top_indices, start=1):
        score = float(scores[idx])
        if score < THRESHOLD:
            break
        row = search_df.iloc[idx]
        results.append({
            "rank"             : rank,
            "question"         : row["question"],
            "answer"           : row["answer"],
            "domain"           : row["domain"],
            "similarity_score" : round(score, 4),
        })
    return results


# ── Session state
if "messages"   not in st.session_state: st.session_state.messages   = []
if "voice_text" not in st.session_state: st.session_state.voice_text = ""


# ── Tabs
tab1, tab2, tab3 = st.tabs(["💬 Chat", "📊 Dataset Explorer", "ℹ️ About"])


# ── TAB 1: Chat
with tab1:
    col_chat, col_side = st.columns([3, 1])

    with col_side:
        st.markdown("### ⚙️ Settings")
        domain_filter = st.selectbox("🏷️ Domain Filter", DOMAINS)
        top_n         = st.slider("🔢 Results to show", 1, 5, 1)
        st.divider()
        st.markdown("### 📈 Stats")
        st.metric("Q&A Pairs", len(df))
        st.metric("Domains",   len(df["domain"].unique()))
        st.metric("Threshold", f"{THRESHOLD}")
        st.divider()
        if st.button("🗑️ Clear Chat", use_container_width=True):
            st.session_state.messages = []
            st.rerun()

    with col_chat:
        st.markdown("## 🤖 Domain Q&A Chatbot")
        st.caption(
            "Retrieval-based chatbot using **TF-IDF + Cosine Similarity**. "
            "Answers come from the dataset — no text is generated."
        )
        st.divider()

        # Welcome message
        if not st.session_state.messages:
            st.markdown(
                "<div class='bot-bubble'>"
                "👋 Hello! Ask me anything about:<br>"
                "<b>AI & ML · Python · Health · Education · Science</b><br><br>"
                "I will find the best matching answer from my knowledge base!"
                "</div>",
                unsafe_allow_html=True,
            )

        # Display messages
        for msg in st.session_state.messages:
            if msg["role"] == "user":
                st.markdown(
                    f"<div class='user-bubble'>👤 {msg['content']}</div>",
                    unsafe_allow_html=True,
                )
            else:
                # Bot message
                answer  = msg["content"]
                matches = msg.get("matches", [])
                st.markdown(
                    f"<div class='bot-bubble'>🤖 {answer}</div>",
                    unsafe_allow_html=True,
                )
                # Show matched questions
                if matches:
                    for m in matches:
                        score_pct = int(m["similarity_score"] * 100)
                        score_bar = "█" * max(1, score_pct // 5)
                        st.markdown(
                            f"<div class='match-box'>"
                            f"📌 Matched: <i>\"{m['question']}\"</i> "
                            f"<span class='domain-badge'>{m['domain']}</span><br>"
                            f"📊 Similarity: {m['similarity_score']:.4f}  "
                            f"<span style='color:#667eea'>{score_bar}</span>"
                            f"</div>",
                            unsafe_allow_html=True,
                        )


        user_input = st.chat_input("Ask your question here...")

        if user_input:
            st.session_state.messages.append(
                {"role": "user", "content": user_input}
            )

            results = get_recommendations(
                user_input,
                domain_filter = domain_filter if domain_filter != "All" else None,
                top_n         = top_n,
            )

            if not results:
                answer  = (
                    "❓ I don't have enough information to answer that.\n\n"
                    "Try rephrasing, or choose a different domain filter."
                )
                matches = []
            else:
                # For top_n=1: show single answer
                # For top_n>1: show numbered answers
                if top_n == 1:
                    answer = results[0]["answer"]
                else:
                    answer = "\n\n".join([
                        f"**{r['rank']}.** {r['answer']}"
                        for r in results
                    ])
                matches = results

            st.session_state.messages.append({
                "role"   : "assistant",
                "content": answer,
                "matches": matches,
            })
            st.rerun()


# ── TAB 2: Dataset Explorer
with tab2:
    st.markdown("## 📊 Dataset Explorer")
    st.markdown(f"**Total Q&A pairs:** {len(df)}")

    # Domain filter
    domain_view = st.selectbox(
        "Filter by domain", ["All"] + sorted(df["domain"].unique().tolist()),
        key="explorer_domain"
    )
    search_term = st.text_input("🔍 Search questions", "")

    view_df = df.copy()
    if domain_view != "All":
        view_df = view_df[view_df["domain"] == domain_view]
    if search_term:
        view_df = view_df[
            view_df["question"].str.lower().str.contains(search_term.lower())
        ]

    st.markdown(f"Showing **{len(view_df)}** entries")
    st.dataframe(
        view_df[["domain","question","answer"]].reset_index(drop=True),
        use_container_width=True,
        height=500,
    )

    # Domain distribution chart
    st.markdown("### Domain Distribution")
    domain_counts = df["domain"].value_counts()
    st.bar_chart(domain_counts)
```
